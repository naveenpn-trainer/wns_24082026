# Prerequisites


```python
%sql
USE CATALOG wns24082026;

DROP TABLE IF EXISTS quickstart_schema.users;

CREATE TABLE quickstart_schema.users (
    id INT,
    name STRING,
    dob DATE,
    email STRING,
    gender STRING,
    country STRING,
    region STRING,
    city STRING,
    asset INT,
    marital_status STRING
  )
```

# Copy Into Using Manual Schema Evolution

## Data Ingestion using Copy Into

```python
%sql
COPY INTO quickstart_schema.users FROM (
  SELECT
    id::INT,
    name::STRING,
    to_date(dob, 'yyyy-MM-dd') as dob,
    email::STRING,
    gender::STRING,
    country::STRING,
    region::STRING,
    city::STRING,
    asset::INT,
    marital_status::STRING FROM
  '/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/input/'
) FILEFORMAT = CSV FORMAT_OPTIONS ('header' = 'true')
```

## Source with more columns

# Copy Into using Automated Schema Evolution

## Prerequisites

```python
%sql
DROP TABLE IF EXISTS quickstart_schema.users;

CREATE TABLE quickstart_schema.users;
DESCRIBE EXTENDED quickstart_schema.users;
```

## Data Ingestion

```python
%sql
COPY INTO quickstart_schema.users FROM
'/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/input/' FILEFORMAT = CSV
FORMAT_OPTIONS (
  'header' = 'true',
  'timestampFormat' = 'yyyy-MM-dd',
  'inferSchema' = 'true',
  'mergeSchema' = 'true'
) COPY_OPTIONS ('mergeSchema' = 'true')
```
