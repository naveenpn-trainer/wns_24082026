# Data Ingestion using Autoloaders

```python
source_configurations = {
    "cloudFiles.format": "csv",
    "header": True,
    "timestampFormat": "yyyy-MM-dd",
    "cloudFiles.schemaLocation": "/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/schema/",
    "cloudFiles.inferColumnTypes": True,
}

destination_configurations = {
    "checkpointLocation": "/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/checkpoint/",
    "mergeSchema": True,
}
source_df = (
    spark.readStream.format("cloudFiles")
    .option("cloudFiles.format", "CSV")
    .options(**source_configurations)
    .load("/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/input")
)

source_df.writeStream.format("DELTA").options(**destination_configurations).outputMode(
    "APPEND"
).trigger(availableNow=True).toTable("quickstart_schema.users")
```
