# Prerequisites

## Create Catalog, and Schema

```python
%sql
USE CATALOG wns24082026;
DROP TABLE IF EXISTS quickstart_schema.users;
```

```python
%sql

DESCRIBE TABLE quickstart_schema.users;
```

## Load 'users_001.csv' into Delta Table

```python
df = spark.read.csv(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/users_001.csv",
    header=True,
    inferSchema=True,
)
df.write.saveAsTable("quickstart_schema.users", mode="OVERWRITE")
```

## Read Delta Table

```python
spark.read.table("quickstart_schema.users").limit(4).display()
```

# Schema Evolution

```python
spark.read.csv(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/users_006_new_column_education.csv",
    header=True,
    inferSchema=True,
).write.option("mergeSchema",True).saveAsTable("quickstart_schema.users", mode="APPEND")
```
