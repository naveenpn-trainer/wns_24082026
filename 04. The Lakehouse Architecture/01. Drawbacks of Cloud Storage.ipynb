# Prerequisites

## Load 'users_001.csv' into DataFrame

```python
df = spark.read.csv(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/user_dataset/users_001.csv",
    header=True,
    inferSchema=True,
)
display(df)
```

## Write as parquet

```python
df.write.save(
    format="PARQUET",
    path="/Volumes/wns24082026/quickstart_schema/sandbox/output_parquet",
)
```

## Read Parquet

```python
df_parquet = spark.read.parquet("/Volumes/wns24082026/quickstart_schema/sandbox/output_parquet")
df_parquet.display()
```

```python
df_parquet.createOrReplaceTempView("users_vw")
```

# [Disadvantage] - Updates are not allowed

```python
%sql
UPDATE
  users_vw
SET
  country = 'Bharat'
where
  country = 'India'
```

# [Disadvantage] - Allows verwrite but no rollback or version history

```python
from pyspark.sql.functions import col

df.filter(col("country") == "India").write.save(
    format="PARQUET",
    path="/Volumes/wns24082026/quickstart_schema/sandbox/output_parquet",
    mode="OVERWRITE",
)
```
