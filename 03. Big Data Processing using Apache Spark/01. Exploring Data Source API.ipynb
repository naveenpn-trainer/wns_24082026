# Creating DataFrame from Different Sources


## Create DataFrame from CSV source

```python
dfr = spark.read
df = dfr.csv(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.csv",
    header=True,
    inferSchema=True
)
display(df)
```

## Create DataFrame from JSON

```python
df = spark.read.json(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users_004.json",
    multiLine=True,
)
display(df)
```

## Create DatFrame from Parquet

```python
df = spark.read.parquet("/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.parquet")
display(df)
```

# Attaching Custom Schema

```python
from pyspark.sql.types import *

USER_SCHEMA = StructType(
    [
        StructField("id", IntegerType()),
        StructField("age", IntegerType()),
        StructField("gender", IntegerType()),
        StructField("designation", StringType()),
        StructField("salary", IntegerType()),
    ]
)

df = spark.read.csv(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.csv",
    header=True,
    schema=USER_SCHEMA,
)
display(df)
```

# Handling Bad Records

1. PERMISSIVE (Default)
2. DROPMALFORMED
3. FAILFAST

```python
spark.read.option("columnNameOfCorruptRecord", "bad_record").json(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/access_logs.json",
    mode="PERMISSIVE",
).display()
```

```python
spark.read.option("columnNameOfCorruptRecord", "bad_record").json(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/access_logs.json",
    mode="DROPMALFORMED",
).display()
```

```python
spark.read.option("columnNameOfCorruptRecord", "bad_record").json(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/access_logs.json",
    mode="FAILFAST",
).display()
```

# DataFrame Writer API

```python
df = spark.read.csv(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.csv",
    header=True,
    inferSchema=True,
)
df.write.save(
    format="PARQUET",
    path="/Volumes/wns24082026/quickstart_schema/sandbox/output/output_parquet",
)
```
