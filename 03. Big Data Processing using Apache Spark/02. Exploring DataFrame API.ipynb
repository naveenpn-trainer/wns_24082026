# Prerequisites

```python
df = spark.read.csv(path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/employee.csv",
                    header=True,
                    inferSchema=True,
                    sep="|",
                    quote="'")
display(df)
```

# Select Operations

```python
df.select("id","name").show()
```

```python
from pyspark.sql.functions import col
df.select(col("id").alias("user_id"), col("name")).show()
```

# Filter Transformation

```python
df.filter((col("desig")=="Developer") | (col("desig")=="R&D Engineer")).show()
```

```python
df.filter(col("desig").isin("Developer","R&D Engineer")).show()
```

```python
from pyspark.sql.functions import lower
df.filter(lower(col("company"))=="cisco").show()
```

# Derive columns

```python
from pyspark.sql.functions import lit
df_new = df.withColumn("category",lit(1))
df_new.printSchema()
```

```python
from pyspark.sql.functions import when, col

result_df = df.withColumn(
    "category",
    when(col("exp") >= 10, "Senior")
    .when(col("exp") >= 5, "Mid Level")
    .when(col("exp") > 0, "Junior")
    .otherwise("Invalid Experience"),
)
result_df.display()
```

```python
result_df.groupBy("category").count().sort("count", ascending=False).display()
```

```python
result_df.groupBy("category").count().sort(col("count").desc()).display()
```

```python
from pyspark.sql.functions import min, max, avg
result_df.groupBy("category").agg(min("exp").alias("min_exp"),
                                  max("exp"),
                                  avg("exp")).show()
```
